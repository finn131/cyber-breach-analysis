# 02 Data Cleaning

Clean the raw dataset and save a reusable processed file.

```python

from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'scripts').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 120)

from scripts.cleaning import clean_data, load_data, save_data

raw_df = load_data()
cleaned_df = clean_data(raw_df)
raw_df.shape, cleaned_df.shape
```

```python
print('Raw missing values:')
print(raw_df.isna().sum())
print('Cleaned missing values:')
print(cleaned_df.isna().sum())
```

```python
print('Raw data types:')
print(raw_df.dtypes)
print('Cleaned data types:')
print(cleaned_df.dtypes)
```

```python
cleaned_df.head()
```

```python
output_path = save_data(cleaned_df)
print(f'Cleaned dataset saved to: {output_path}')
```


## Cleaning Notes

- The leading CSV index column is removed.
- Text columns are trimmed and lowercased.
- Date fields are parsed into datetime values.
- `PwnCount` is converted to numeric.
- Boolean flag columns are normalized to nullable booleans.
- Duplicate rows are removed.
